Question: does DINO similarity separate same and different manga characters? No clustering. Human labels are needed to answer this.

```python
from pathlib import Path
import sys, json
from datetime import datetime
from IPython.display import display, Image as DisplayImage
ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
if ROOT.name != 'DINO-EXPERMINTATOIN-IDENTITY':
    ROOT = ROOT / 'DINO-EXPERMINTATOIN-IDENTITY'
assert (ROOT / 'detection.py').exists(), ROOT
sys.path.insert(0, str(ROOT))
import automatic_identity as auto
```

```python
DETECTIONS = r"c:\Users\ESHAAN\BYTE\Applied-Task\MultiModelSystem\DINO-EXPERMINTATOIN-IDENTITY\outputs\detection_yolov3_20261008_060537_609792\detections.json"  # Set Path(r'C:/.../detections.json')
assert DETECTIONS is not None, 'Paste the manifest path from notebook 02'
THRESHOLD = 0.8  # exploratory; not calibrated
output = ROOT / 'outputs' / f'pairs_{datetime.now():%Y%m%d_%H%M%S_%f}'
summary = auto.run_identity(DETECTIONS, output, threshold=THRESHOLD)
print(summary)
print(output / 'pair_review.html')
```
