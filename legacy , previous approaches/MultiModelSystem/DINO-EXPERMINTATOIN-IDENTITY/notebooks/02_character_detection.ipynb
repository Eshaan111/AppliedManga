# 1. Detect faces and crop them

Only hysts YOLOv3 face detection. No landmarks. Run download_detector.py once before this notebook. Manual annotations are preserved.

```python
from pathlib import Path
import sys, json
from datetime import datetime
from IPython.display import display, Image as DisplayImage
ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
if ROOT.name != 'DINO-EXPERMINTATOIN-IDENTITY':
    ROOT = ROOT / 'DINO-EXPERMINTATOIN-IDENTITY'
assert (ROOT / 'detection.py').exists(), ROOT
sys.path.insert(0, str(ROOT))
import detection
```

```python
PAGES = 3
CONFIDENCE = 0.5
pages = detection.select_pages(limit=PAGES)
output = ROOT / 'outputs' / f'detection_yolov3_{datetime.now():%Y%m%d_%H%M%S_%f}'
manifest = detection.run_detection(pages, output, confidence=CONFIDENCE)
print('Use this in notebook 03:', output / 'detections.json')
```
