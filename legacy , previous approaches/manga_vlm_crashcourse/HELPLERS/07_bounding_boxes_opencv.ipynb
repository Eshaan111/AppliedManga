## Bounding-box helpers

Use a consistent convention such as `[x1, y1, x2, y2]` and never mix it with `[x, y, width, height]` silently.

```python
def center(box):
    x1, y1, x2, y2 = box
    return ((x1 + x2) / 2, (y1 + y2) / 2)

def area(box):
    x1, y1, x2, y2 = box
    return max(0, x2-x1) * max(0, y2-y1)

print(center([10, 20, 110, 80]))
print(area([10, 20, 110, 80]))
```

## IoU

Intersection-over-Union is fundamental for comparing detections and measuring overlap.

```python
def iou(a, b):
    ax1, ay1, ax2, ay2 = a
    bx1, by1, bx2, by2 = b
    ix1, iy1 = max(ax1,bx1), max(ay1,by1)
    ix2, iy2 = min(ax2,bx2), min(ay2,by2)
    inter = max(0, ix2-ix1) * max(0, iy2-iy1)
    union = area(a) + area(b) - inter
    return inter / union if union else 0.0

print(iou([0,0,100,100], [50,50,150,150]))
```

## Visual debugging

Draw every OCR box and assign an index. If ordering is wrong, you should be able to see it immediately on the page.

```python
def draw_boxes(image, items):
    img = image.copy()
    for i, item in enumerate(items):
        x1, y1, x2, y2 = map(int, item["bbox"])
        cv2.rectangle(img, (x1,y1), (x2,y2), (255,255,255), 2)
        cv2.putText(img, str(i), (x1, max(0,y1-5)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 2)
    return img
```
