## Ablation thinking

Ablations answer *why* a system improved.

Examples:
- without LoRA;
- without OCR preprocessing;
- without panel ordering;
- without cross-page identity matching;
- single page vs three pages;
- low vs high image resolution.

Do not change five components simultaneously and then claim you know what helped.

## Suggested development loop

1. Establish a reproducible baseline.
2. Save dev predictions.
3. Run the official scorer.
4. Inspect 10–20 failures.
5. Count failure categories.
6. Choose the dominant failure mode.
7. Modify only the relevant subsystem.
8. Re-run and compare.
9. Keep both successful and failed experiments in your notes.
