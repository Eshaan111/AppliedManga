## Patchification

A ViT splits an image into fixed-size patches. A `224×224` image with `16×16` patches yields `14×14 = 196` patch tokens.

Higher image resolution produces more patches and therefore more compute, but can preserve tiny text better.

```python
def num_patches(h, w, patch):
    assert h % patch == 0 and w % patch == 0
    return (h // patch) * (w // patch)

for res in [224, 336, 448, 896]:
    print(res, "->", num_patches(res, res, 16), "patches")
```

## Simple patchify demonstration

This exercise shows what patchification does to tensor shapes.

```python
import torch

x = torch.randn(1, 3, 224, 224)
patch = 16

patches = x.unfold(2, patch, patch).unfold(3, patch, patch)
print("After unfold:", patches.shape)

patches = patches.permute(0, 2, 3, 1, 4, 5).contiguous()
patches = patches.view(1, -1, 3 * patch * patch)
print("Token-like patches:", patches.shape)
```
