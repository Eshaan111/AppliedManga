## Embedding intuition

An image encoder maps a crop to a feature vector. Similar visual identity should ideally produce nearby vectors even when pose changes.

CLIP embeddings are trained around image-text alignment. DINO-style features are self-supervised visual representations and can be strong for appearance similarity.

## Cosine similarity

Cosine similarity compares vector direction rather than raw magnitude.

```python
import torch
import torch.nn.functional as F

a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([1.1, 1.9, 3.2])

sim = F.cosine_similarity(a.unsqueeze(0), b.unsqueeze(0))
print(float(sim))
```

## Batch-normalized similarity matrix

Useful when matching several candidate characters between pages.

```python
emb = torch.randn(6, 768)
emb = F.normalize(emb, dim=-1)
similarity_matrix = emb @ emb.T
print(similarity_matrix.shape)
```
