## Core architecture

Most VLMs contain a vision encoder, a bridge/projector, and a language model. The image becomes a sequence of visual embeddings that the language model can condition on.

General flow:

`image → vision encoder → visual tokens → projector/resampler → language-model embedding space → autoregressive decoder`

```python
# Shape thought experiment
B = 2
num_visual_tokens = 576
vision_dim = 1024
lm_dim = 2048

print("Vision features:", (B, num_visual_tokens, vision_dim))
print("Projected features:", (B, num_visual_tokens, lm_dim))
```

## Fusion strategies

Common designs include projection of patch tokens, query-based resamplers, cross-attention, and inserting image tokens directly into the decoder sequence.

The exact mechanism matters when deciding what to fine-tune and how much memory inference uses.

## Why manga is difficult

Manga simultaneously demands tiny-text recognition, unusual layouts, stylized line art, speaker grounding, and cross-page identity. These are not the same capability.

A model can transcribe text well while failing reading order. It can also understand story context while hallucinating text.
