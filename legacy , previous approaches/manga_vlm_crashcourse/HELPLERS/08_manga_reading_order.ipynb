## Why naïve sorting fails

For simple layouts, top-to-bottom and right-to-left may work. Complex layouts have nested panels, overlapping balloons, and nonuniform panel heights.

Treat reading order as a structural problem, not merely a coordinate sort.

```python
items = [
    {"id":"A", "bbox":[500, 50, 700, 160]},
    {"id":"B", "bbox":[100, 60, 300, 170]},
    {"id":"C", "bbox":[420, 250, 620, 360]},
    {"id":"D", "bbox":[120, 260, 320, 370]},
]

# Crude baseline only:
ordered = sorted(items, key=lambda x: (center(x["bbox"])[1], -center(x["bbox"])[0]))
print([x["id"] for x in ordered])
```

## Panel-first strategy

One robust decomposition is:

`detect panels → order panels → order bubbles inside each panel`

Even if panel detection is heuristic, it gives you a place to encode manga layout structure.

## Graph formulation

You can model candidate precedence relations as a directed graph. For example, object A precedes B if A lies clearly above B, or if they share a row and A is to the right.

The graph can later be topologically sorted if it is acyclic.

```python
import networkx as nx

G = nx.DiGraph()
G.add_edges_from([("A","B"), ("A","C"), ("B","D"), ("C","D")])
print(list(nx.topological_sort(G)))
```
