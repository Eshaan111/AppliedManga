## OCR decomposition

OCR is usually at least two problems:

1. detect text regions;
2. recognize characters inside each region.

A manga pipeline may then add filtering and reading-order logic.

## Image preprocessing playground

Thresholding can improve some scans and destroy others. Treat preprocessing as an experiment rather than a universal rule.

```python
import cv2
import numpy as np

def preprocess(gray):
    if gray.ndim == 3:
        gray = cv2.cvtColor(gray, cv2.COLOR_BGR2GRAY)
    binary = cv2.adaptiveThreshold(
        gray, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        31, 11
    )
    return binary
```

## Manga-specific concerns

Watch for vertical text, stylized fonts, outlined lettering, curved balloons, punctuation-only balloons, tiny reactions, and false positives from sound effects.

A useful experiment is precision-first filtering vs recall-first extraction, because hallucinated/irrelevant text is penalized.
