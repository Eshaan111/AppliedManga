Inspect the current kernel, installed packages, model paths and raw OCR caches.


```python
# Cell purpose: Inspect the current kernel, installed packages, model paths and raw OCR caches.
# Inputs: Local environment and the configured data/cache directories.
# Outputs: Printed checks and cache/discovery.json; no model inference or training.

import sys
import importlib.metadata

report = {'python': sys.executable, 'paths': {}, 'raw_cache_counts': {}}
print('Kernel Python:', sys.executable)
for package in ['torch', 'transformers', 'numpy', 'Pillow', 'sentencepiece']:
    try:
        print(package, importlib.metadata.version(package))
    except importlib.metadata.PackageNotFoundError:
        print(package, 'MISSING — install in this kernel')
for key in ['labels_jsonl', 'images_root', 'magi_model', 'dino_model', 'legacy_visual_cache']:
    value = CONFIG[key]
    exists = bool(value and Path(value).exists())
    report['paths'][key] = {'path': str(value), 'exists': exists}
    print(key, ':', value, '| exists:', exists)
for directory in CONFIG['raw_cache_roots']:
    files = sorted(directory.glob('*/magi_raw.json'))
    report['raw_cache_counts'][str(directory)] = len(files)
    print('Raw Magi/OCR:', directory, '| sequences:', len(files))
    if files:
        example = json.loads(files[0].read_text(encoding='utf-8'))
        print('Example page fields:', list(example['data'][0]))
save_json(CACHE / 'discovery.json', report)
```
